# 01 — Exploratory Data Analysis (EDA)
### VoltIQ: EV Charging Demand Forecasting & Station Intelligence
Analyzes 50,000 historical charging events across 30 charging stations to uncover diurnal, seasonal, and station-type demand distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned sessions
sessions = pd.read_csv('../data/interim/sessions_clean.csv', parse_dates=['start_time', 'end_time'])
print(f"Loaded {len(sessions):,} cleaned sessions across {sessions['station_id'].nunique()} stations.")
sessions.head()
```

```python
# Energy Delivery Distribution by Station Type
plt.figure(figsize=(10, 5))
sns.boxplot(data=sessions, x='station_type', y='energy_kwh', palette='viridis')
plt.title('Energy Delivered (kWh) by Charger Category (L1 vs L2 vs DC Fast)')
plt.xlabel('Charger Class')
plt.ylabel('Energy Delivered (kWh)')
plt.grid(True, alpha=0.3)
plt.show()
```
