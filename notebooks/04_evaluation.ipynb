# 04 — Model Evaluation & Algorithmic Fairness
### VoltIQ Fairness & Disparity Audit
Validating error rates across Urban, Suburban, Highway, and Rural stations to ensure equitable capacity allocation.

```python
import json
from voltiq.config.settings import settings

summary_path = settings.models_store_dir / 'pipeline_summary.json'
with open(summary_path) as f:
    summary = json.load(f)

print("Champion Models:")
print(f"Best Regressor: {summary['regression_leaderboard'][0]['model']} (RMSE: {summary['regression_leaderboard'][0]['rmse']})")
print(f"Best Classifier: {summary['classification_leaderboard'][0]['model']} (F1: {summary['classification_leaderboard'][0]['f1_macro']})")

print("\nSpatial Fairness by Cohort:")
for loc, metrics in summary['fairness_by_location'].items():
    print(f"- {loc:10s} | MAE: {metrics['mae']:.4f} | Relative MAE: {metrics['relative_mae']*100:.1f}%")
```
