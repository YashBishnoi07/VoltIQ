# 02 — Statistical Hypothesis Testing
### VoltIQ Statistical Rigor Suite
Empirical testing of normality (Shapiro-Wilk / KS), Mann-Whitney U test for weekend effect, and Kruskal-Wallis ANOVA for peak hour concurrency.

```python
import pandas as pd
from voltiq.analysis.statistical_tests import test_normality, test_weekday_vs_weekend, test_peak_hours_anova

hourly = pd.read_csv('../data/processed/hourly_features.csv', parse_dates=['hour_bucket'])

# 1. Normality test
norm = test_normality(hourly['sessions_per_hour'], 'sessions_per_hour')
print("Normality Test:", norm)

# 2. Weekday vs Weekend (Mann-Whitney U)
ww = test_weekday_vs_weekend(hourly)
print("\nWeekday vs Weekend Test:", ww)

# 3. Peak Hour Kruskal-Wallis ANOVA
anova = test_peak_hours_anova(hourly)
print("\nHour of Day ANOVA:", anova)
```
